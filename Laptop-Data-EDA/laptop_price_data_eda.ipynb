### 4. **Correcting data types**

Change the types of all size columns

```python
print(df['ram_gb'].value_counts())
print(df['ssd'].value_counts())
print(df['hdd'].value_counts())
```

```python
df['ram_gb']=df['ram_gb'].apply(lambda x: x.replace('GB', '')if 'GB' in str(x) else(x))
df['ssd']=df['ssd'].apply(lambda x: x.replace('GB', '')if 'GB' in str(x) else(x))
df['hdd']=df['hdd'].apply(lambda x: x.replace('GB', '')if 'GB' in str(x) else(x))
df['graphic_card_gb']=df['graphic_card_gb'].apply(lambda x: x.replace('GB', '')if 'GB' in str(x) else(x))
df['os_bit']=df['os_bit'].apply(lambda x: x.replace('-bit', '')if '-bit' in str(x) else(x))
df['rating']=df['rating'].apply(lambda x: x.replace('stars', '')if 'stars' in str(x) else 'star' in str(x))
df.head()
```

```python
df.info()
```

```python
#rename some columns
df.rename(columns={'brand':'Brand'}, inplace=True)
df.rename(columns={'processor_brand':'Processor_Brand'}, inplace=True)
df.rename(columns={'ram_gb':'RAM(GB)'}, inplace=True)
df.rename(columns={'ssd':'SSD(GB)'}, inplace=True)
df.rename(columns={'hdd':'HDD(GB)'}, inplace=True)
df.rename(columns={'graphic_card_gb':'Graphic card(GB)'}, inplace=True)
df.head()
```

```python
df['RAM(GB)']=pd.to_numeric(df['RAM(GB)'], errors='coerce')
df['SSD(GB)']=pd.to_numeric(df['SSD(GB)'], errors='coerce')
df['HDD(GB)']=pd.to_numeric(df['HDD(GB)'], errors='coerce')
df['Graphic card(GB)']=pd.to_numeric(df['Graphic card(GB)'], errors='coerce')
df['os_bit']=pd.to_numeric(df['os_bit'], errors='coerce')
df['rating']=pd.to_numeric(df['rating'], errors='coerce')
df.info()
```

**Observation**
- Following types are converted in numeric.
- Ram,ssd,hdd,grpahic card, os-bit, rating
